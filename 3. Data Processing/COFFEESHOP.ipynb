```sql
SELECT *
FROM `bright-tv`.default.brightcoffeeshop;

SELECT DISTINCT product_type
FROM `bright-tv`.default.brightcoffeeshop;

SELECT DISTINCT
       product_type,
     CASE 
         WHEN product_type IS NULL THEN 'Unknown' 
         WHEN TRIM(product_type) = '' THEN 'Unknown'
         ELSE product_type
    END AS product_typ
FROM `bright-tv`.default.brightcoffeeshop;

SELECT transaction_id,
       COUNT(*) AS Duplicate_cnt
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY transaction_id
HAVING COUNT(*) > 1
ORDER BY Duplicate_cnt DESC;

SELECT
    product_type,
    ROUND(SUM(Unit_price_standardized * transaction_qty), 2) AS total_revenue
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY product_type
ORDER BY total_revenue DESC;

SELECT
    product_detail,
    SUM(transaction_qty) AS total_units_sold
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY product_detail
ORDER BY total_units_sold DESC;

SELECT DISTINCT Unit_price_standardized
FROM  `bright-tv`.default.brightcoffeeshop
ORDER BY Unit_price_standardized;

SELECT *
FROM  `bright-tv`.default.brightcoffeeshop
WHERE transaction_qty <= 0
   OR Unit_price_standardized IS NULL;

SELECT
    CAST(REPLACE(Unit_price_standardized, ',', '.') AS DECIMAL(10,2)) AS unit_price
FROM  `bright-tv`.default.brightcoffeeshop;

SELECT *,
       Unit_price_standardized * transaction_qty AS total_amount
FROM `bright-tv`.default.brightcoffeeshop;

SELECT transaction_qty,
       Unit_price_standardized,
       ROUND(SUM(Unit_price_standardized * transaction_qty), 2) AS Total_Amount
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY transaction_qty, Unit_price_standardized;

SELECT transaction_qty, ROUND(SUM(CAST(transaction_qty AS DOUBLE)* CAST(Unit_price_standardized AS DOUBLE)),0) AS Total_Amount
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY transaction_qty;

SELECT SUM(Unit_price_standardized * transaction_qty) AS total_revenue
FROM `bright-tv`.default.brightcoffeeshop;

SELECT product_category,
    SUM(Unit_price_standardized * transaction_qty) AS total_revenue
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY product_category
ORDER BY total_revenue DESC;


SELECT 
    product_detail,
    SUM(Unit_price_standardized * transaction_qty) AS total_revenue
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY product_detail
ORDER BY total_revenue DESC;

SELECT 
    store_location,
    SUM(Unit_price_standardized * transaction_qty) AS total_revenue
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY store_location
ORDER BY total_revenue DESC;

SELECT 
    product_category,
    SUM(transaction_qty) AS total_items_sold,
    SUM(Unit_price_standardized * transaction_qty) AS total_revenue
FROM `bright-tv`.default.brightcoffeeshop
GROUP BY product_category
ORDER BY total_revenue DESC;
```
